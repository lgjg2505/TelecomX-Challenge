```python
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

with open('TelecomX_Data.json', 'r', encoding='utf-8') as file:
    raw_data = json.load(file)

df = pd.json_normalize(raw_data)

display(df.head())
```

```python
df.columns = df.columns.str.replace(r'^(customer|phone|internet|account)\.', '', regex=True)

df['Charges.Total'] = pd.to_numeric(df['Charges.Total'], errors='coerce')

df['Charges.Total'] = df['Charges.Total'].fillna(0)

print(df.info())
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x='Churn', palette='pastel')
plt.title('Distribución de Evasión de Clientes (Churn)', fontsize=14)
plt.xlabel('¿Dejó la empresa? (Churn)', fontsize=12)
plt.ylabel('Cantidad de Clientes', fontsize=12)
for p in ax.patches:
    ax.annotate(f'{p.get_height()}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')
plt.show()

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Contract', hue='Churn', palette='Set2')
plt.title('Evasión de Clientes según Tipo de Contrato', fontsize=14)
plt.xlabel('Tipo de Contrato', fontsize=12)
plt.ylabel('Cantidad de Clientes', fontsize=12)
plt.legend(title='Churn')
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Churn', y='Charges.Monthly', palette='Set3')
plt.title('Distribución de Cargos Mensuales en relación a la Evasión', fontsize=14)
plt.xlabel('¿Dejó la empresa? (Churn)', fontsize=12)
plt.ylabel('Cargos Mensuales ($)', fontsize=12)
plt.show()
```
